## 1. Data Preparation

The first major task is to access and evaluate the data.  
The dataset comes from multiple sources and contains information about user subscriptions and streaming activities.

### **Datasets**

- **User Profile data – `members.csv`**
  Includes user demographic details:
  - Age
  - City
  - Registration date/time

- **User Logs data – `user_logs.csv`**
  Captures each user's streaming behavior:
  - Songs played per day
  - Playback activity history

- **User Transaction data – `transactions.csv`**
  Contains subscription-related information:
  - Payment method
  - Subscription renewal / cancellation status

- **Historical / Label data – `train.csv`**
  Includes:
  - `user_id`
  - `is_churn` (target variable indicating churn or not)
